```python
import numpy as np
import matplotlib.pyplot as plt


clean_curves = []
mislabeled_curves = []

epochs = range(1, raw_data["epoch"].max()+2, 1)

for _, filtered_df in raw_data.groupby("id"):
    label = filtered_df["label"].iloc[0]
    metric = f"c{label}_grad_norm"
    curve = filtered_df[metric].values

    mislabeled = int(filtered_df["id"].iloc[0]) in diff_indices  

    if mislabeled:
        mislabeled_curves.append(curve)
    else:
        clean_curves.append(curve)

clean_curves = np.array(clean_curves)
mislabeled_curves = np.array(mislabeled_curves)

avg_clean = np.median(clean_curves, axis=0)

avg_mislabeled = np.median(mislabeled_curves, axis=0)

plt.figure(figsize=(8, 6))

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 12
})

plt.plot(epochs, avg_clean, label="Clean Samples", color="green", linewidth=2)

plt.plot(epochs, avg_mislabeled, label="Mislabeled Samples", color="red", linewidth=2)


plt.xlabel("Epoch")
plt.ylabel("Median Gradient Norm")
#plt.title("Median Gradient Norm Curves")
plt.legend(frameon=True, loc="upper right")
plt.grid(False)

save_dir = "thesis images"
os.makedirs(save_dir, exist_ok=True)  
save_path = os.path.join(save_dir, "cifar10_median_grad_norm_mislabel_vs_clean.png")

# Save the figure in full resolution
#plt.savefig(save_path, bbox_inches='tight', dpi=600)


plt.show()


    
```

```python
from sklearn.metrics import accuracy_score, recall_score, f1_score
import pandas as pd

metrics = []

for i in range(101):
    fraction = i / 100

    # --- Berechnung für count df ---
    count_threshold = count_df['lowest_count_label'].quantile(fraction)
    
    mislabeles_count = count_df[count_df['lowest_count_label'] <= count_threshold]
    non_mislabeles_count = count_df[count_df['lowest_count_label'] > count_threshold]
    
    y_true_count = count_df['mislabeled']
    y_pred_count = [0 if id_ in non_mislabeles_count['id'].values else 1 for id_ in count_df['id']] 
    
    accuracy_count = accuracy_score(y_true_count, y_pred_count)
    recall_count = recall_score(y_true_count, y_pred_count, zero_division=0)
    f1_count = f1_score(y_true_count, y_pred_count, zero_division=0)
    
    metrics.append({
        'Fraction': fraction,
        'Accuracy': accuracy_count,
        'Recall': recall_count,
        'F1': f1_count,
        'Method': 'Count'
    })

    # --- Berechnung für count star df ---
    count_threshold = count_star_df['lowest_count_label'].quantile(fraction)
    
    mislabeles_count = count_star_df[count_star_df['lowest_count_label'] <= count_threshold]
    non_mislabeles_count = count_star_df[count_star_df['lowest_count_label'] > count_threshold]
    
    y_true_count = count_star_df['mislabeled']
    y_pred_count = [0 if id_ in non_mislabeles_count['id'].values else 1 for id_ in count_star_df['id']] 
    
    accuracy_count = accuracy_score(y_true_count, y_pred_count)
    recall_count = recall_score(y_true_count, y_pred_count, zero_division=0)
    f1_count = f1_score(y_true_count, y_pred_count, zero_division=0)
    
    metrics.append({
        'Fraction': fraction,
        'Accuracy': accuracy_count,
        'Recall': recall_count,
        'F1': f1_count,
        'Method': 'Count*'
    })
    
    # --- Berechnung für tracin_df ---
    tracin_threshold = tracin_df['tracin'].quantile(1 - fraction)
    
    filtered_tracin = tracin_df[tracin_df['tracin'] <= tracin_threshold]
    
    y_true_tracin = tracin_df['mislabeled']
    y_pred_tracin = [0 if id_ in filtered_tracin['id'].values else 1 for id_ in tracin_df['id']]
    
    accuracy_tracin = accuracy_score(y_true_tracin, y_pred_tracin)
    recall_tracin = recall_score(y_true_tracin, y_pred_tracin, zero_division=0)
    f1_tracin = f1_score(y_true_tracin, y_pred_tracin, zero_division=0)
    
    metrics.append({
        'Fraction': fraction,
        'Accuracy': accuracy_tracin,
        'Recall': recall_tracin,
        'F1': f1_tracin,
        'Method': 'TracIn'
    })

    # --- Berechnung für distance_df ---
    distance_threshold = distance_df['distance'].quantile(1 - fraction)
    
    filtered_distance = distance_df[distance_df['distance'] <= distance_threshold]
    
    y_true_distance = distance_df['mislabeled']
    y_pred_distance = [0 if id_ in filtered_distance['id'].values else 1 for id_ in distance_df['id']]
    
    accuracy_distance = accuracy_score(y_true_distance, y_pred_distance)
    recall_distance = recall_score(y_true_distance, y_pred_distance, zero_division=0)
    f1_distance = f1_score(y_true_distance, y_pred_distance, zero_division=0)
    
    metrics.append({
        'Fraction': fraction,
        'Accuracy': accuracy_distance,
        'Recall': recall_distance,
        'F1': f1_distance,
        'Method': 'Null Line'
    })

metrics_df = pd.DataFrame(metrics)

metrics_melted = metrics_df.melt(
    id_vars=["Fraction", "Method"], 
    value_vars=["Accuracy", "Recall", "F1"],
    var_name="Metric", 
    value_name="Value"
)
```

# Random Stuff

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Define loss surfaces
def loss_surface1(w1, w2):
    return np.sin(w1) * np.cos(w2) + 0.1 * (w1**2 + w2**2)

def loss_surface2(w1, w2):
    return np.cos(w1) * np.sin(w2) + 0.1 * (w1**2 + w2**2)

# Meshgrid
w1 = np.linspace(-3, 3, 100)
w2 = np.linspace(-3, 3, 100)
W1, W2 = np.meshgrid(w1, w2)
Z1 = loss_surface1(W1, W2)
Z2 = loss_surface2(W1, W2)

# Gradient path function
def gradient_path(loss_func, w_init, steps=5, lr=0.2):
    path = [w_init]
    w = np.array(w_init, dtype=float)
    for _ in range(steps):
        dw1 = (loss_func(w[0] + 1e-4, w[1]) - loss_func(w[0], w[1])) / 1e-4
        dw2 = (loss_func(w[0], w[1] + 1e-4) - loss_func(w[0], w[1])) / 1e-4
        grad = np.array([dw1, dw2])
        w -= lr * grad
        path.append(w.copy())
    return np.array(path)

path1 = gradient_path(loss_surface1, [-2, 2])
path2 = gradient_path(loss_surface2, [2, -2])

# Plot
fig = plt.figure(figsize=(10, 5))
plt.subplots_adjust(wspace=0.1)

# Plot 1
ax1 = fig.add_subplot(121, projection='3d')
ax1.plot_surface(W1, W2, Z1, cmap='coolwarm', edgecolor='k', linewidth=0.3, alpha=0.9, rstride=5, cstride=5)
ax1.contour(W1, W2, Z1, 10, offset=np.min(Z1)-1, cmap='coolwarm', linestyles="solid")

ax1.set_title(r"\textbf{Sample $x_1$}", fontsize=13, y=1.02)
ax1.set_xlabel(r"$w_1$", labelpad=10)
ax1.set_ylabel(r"$w_2$", labelpad=10)
ax1.set_zlabel(r"$\mathcal{L}(y_1, \hat{y}_1)$", labelpad=10)

# Gradient arrows
for i in range(len(path1) - 1):
    x, y = path1[i]
    z = loss_surface1(x, y)
    dx, dy = path1[i+1] - path1[i]
    dz = loss_surface1(*path1[i+1]) - z
    ax1.quiver(x, y, z, dx, dy, dz, color='green', linewidth=2, arrow_length_ratio=0.15)

# Plot 2
ax2 = fig.add_subplot(122, projection='3d')
ax2.plot_surface(W1, W2, Z2, cmap='coolwarm', edgecolor='k', linewidth=0.3, alpha=0.9, rstride=5, cstride=5)
ax2.contour(W1, W2, Z2, 10, offset=np.min(Z2)-1, cmap='coolwarm', linestyles="solid")

ax2.set_title(r"\textbf{Sample $x_2$}", fontsize=13, y=1.02)
ax2.set_xlabel(r"$w_1$", labelpad=10)
ax2.set_ylabel(r"$w_2$", labelpad=10)
ax2.set_zlabel(r"$\mathcal{L}(y_2, \hat{y}_2)$", labelpad=10)

for i in range(len(path2) - 1):
    x, y = path2[i]
    z = loss_surface2(x, y)
    dx, dy = path2[i+1] - path2[i]
    dz = loss_surface2(*path2[i+1]) - z
    ax2.quiver(x, y, z, dx, dy, dz, color='purple', linewidth=2, arrow_length_ratio=0.15)

# Make it beautiful
for ax in [ax1, ax2]:
    ax.view_init(elev=35, azim=-60)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_zticks([])
    ax.grid(False)

plt.show()

```

```python
import pandas as pd

count_df = pd.read_pickle(f'{directory}/counts_after_thesis.pkl')
```

```python
# New combined score

count_df['rank_glarex_score'] = count_df['glarex score'].rank(ascending=True, method='average')
count_df['rank_grad_norm_sum'] = count_df['grad norm sum'].rank(ascending=False, method='average')
count_df['rank_distance'] = count_df['distance'].rank(ascending=True, method='average')
count_df['rank_sections'] = count_df['sections'].rank(ascending=False, method='average')

# Step 2: Calculate the average of the four ranks
count_df['combined rank'] = count_df[['rank_glarex_score', 'rank_grad_norm_sum', 'rank_distance', 'rank_sections']].mean(axis=1)
```

```python
from sklearn.metrics import accuracy_score, recall_score, f1_score
import pandas as pd

metrics = []

for i in range(101):
    fraction = i / 100

    # --- Berechnung für glarex ---
    count_threshold = count_df['glarex score'].quantile(fraction)
    
    mislabeles_count = count_df[count_df['glarex score'] <= count_threshold]
    non_mislabeles_count = count_df[count_df['glarex score'] > count_threshold]
    
    y_true_count = count_df['mislabeled']
    y_pred_count = [0 if id_ in non_mislabeles_count['id'].values else 1 for id_ in count_df['id']] 
    
    accuracy_count = accuracy_score(y_true_count, y_pred_count)
    recall_count = recall_score(y_true_count, y_pred_count, zero_division=0)
    f1_count = f1_score(y_true_count, y_pred_count, zero_division=0)
    
    metrics.append({
        'Fraction': fraction,
        'Accuracy': accuracy_count,
        'Recall': recall_count,
        'F1': f1_count,
        'Method': 'GLAREx'
    })

    # --- Berechnung für grad norm sum ---
    count_threshold = count_df['grad norm sum'].quantile(1-fraction)
    
    mislabeles_count = count_df[count_df['grad norm sum'] >= count_threshold]
    non_mislabeles_count = count_df[count_df['grad norm sum'] < count_threshold]
    
    y_true_count = count_df['mislabeled']
    y_pred_count = [0 if id_ in non_mislabeles_count['id'].values else 1 for id_ in count_df['id']] 
    
    accuracy_count = accuracy_score(y_true_count, y_pred_count)
    recall_count = recall_score(y_true_count, y_pred_count, zero_division=0)
    f1_count = f1_score(y_true_count, y_pred_count, zero_division=0)
    
    metrics.append({
        'Fraction': fraction,
        'Accuracy': accuracy_count,
        'Recall': recall_count,
        'F1': f1_count,
        'Method': 'Grad Norm Sum'
    })
    
    # --- Berechnung für distance ---
    count_threshold = count_df['distance'].quantile(fraction)
    
    mislabeles_count = count_df[count_df['distance'] <= count_threshold]
    non_mislabeles_count = count_df[count_df['distance'] > count_threshold]
    
    y_true_count = count_df['mislabeled']
    y_pred_count = [0 if id_ in non_mislabeles_count['id'].values else 1 for id_ in count_df['id']] 
    
    accuracy_count = accuracy_score(y_true_count, y_pred_count)
    recall_count = recall_score(y_true_count, y_pred_count, zero_division=0)
    f1_count = f1_score(y_true_count, y_pred_count, zero_division=0)
    
    metrics.append({
        'Fraction': fraction,
        'Accuracy': accuracy_count,
        'Recall': recall_count,
        'F1': f1_count,
        'Method': 'Distance'
    })

    # --- Berechnung für sections ---
    count_threshold = count_df['sections'].quantile(1-fraction)
    
    mislabeles_count = count_df[count_df['sections'] >= count_threshold]
    non_mislabeles_count = count_df[count_df['sections'] < count_threshold]
    
    y_true_count = count_df['mislabeled']
    y_pred_count = [0 if id_ in non_mislabeles_count['id'].values else 1 for id_ in count_df['id']] 
    
    accuracy_count = accuracy_score(y_true_count, y_pred_count)
    recall_count = recall_score(y_true_count, y_pred_count, zero_division=0)
    f1_count = f1_score(y_true_count, y_pred_count, zero_division=0)
    
    metrics.append({
        'Fraction': fraction,
        'Accuracy': accuracy_count,
        'Recall': recall_count,
        'F1': f1_count,
        'Method': 'Sections'
    })


    # --- Berechnung für combined rank ---
    count_threshold = count_df['combined rank'].quantile(fraction)
    
    mislabeles_count = count_df[count_df['combined rank'] <= count_threshold]
    non_mislabeles_count = count_df[count_df['combined rank'] > count_threshold]
    
    y_true_count = count_df['mislabeled']
    y_pred_count = [0 if id_ in non_mislabeles_count['id'].values else 1 for id_ in count_df['id']] 
    
    accuracy_count = accuracy_score(y_true_count, y_pred_count)
    recall_count = recall_score(y_true_count, y_pred_count, zero_division=0)
    f1_count = f1_score(y_true_count, y_pred_count, zero_division=0)
    
    metrics.append({
        'Fraction': fraction,
        'Accuracy': accuracy_count,
        'Recall': recall_count,
        'F1': f1_count,
        'Method': 'Combined'
    })

metrics_df = pd.DataFrame(metrics)

metrics_melted = metrics_df.melt(
    id_vars=["Fraction", "Method"], 
    value_vars=["Accuracy", "Recall", "F1"],
    var_name="Metric", 
    value_name="Value"
)
```
